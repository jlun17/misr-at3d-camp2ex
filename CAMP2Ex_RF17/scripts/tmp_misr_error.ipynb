```python
import numpy as np
```

```python
costs = np.zeros(100)
for i in range(len(costs)):
    n = 3000
    sigma = 2
    y = np.arange(n)
    F = np.arange(n) + np.random.normal(loc=0.0, scale=sigma, size=n)
    error_cov = np.diag(sigma*np.ones(n))
    inv_error_cov = np.linalg.inv(error_cov)
    costs[i] = (1/(2*n)) * np.matmul(y-F, np.matmul(inv_error_cov, y-F))
```

```python
import matplotlib.pyplot as plt

f, ax = plt.subplots(1, 1, figsize=(8, 5))
h = ax.hist(costs, bins=10)
```
