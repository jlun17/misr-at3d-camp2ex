```python
F_sensor = load_sensor_dict(iterfile)
y_sensor = load_sensor_dict(obs_file)
```

```python
def obs_vec(sensor):
    vec = np.array([])
    for s in sensor['MISR']['sensor_list']:
        # condx = (s.ray_x.data >= 0.0) & (s.ray_x.data <= 17.6)
        # condy = (s.ray_y.data >= 0.0) & (s.ray_y.data <= 17.6)
        # cond = condx & condy
        # vec = np.append(vec, (np.pi * s.I.data[cond]) / s.mu0)
        vec = np.append(vec, (np.pi * s.I.data) / s.mu0)
    return vec
```

```python
F = obs_vec(F_sensor)
y = obs_vec(y_sensor)

S_1 = np.diag(1 / (0.04 * y)**2)
chi2 = (1/len(y)) * np.matmul(y-F, np.matmul(S_1, y-F))

print('chi^2 = {:.2f}'.format(chi2))
```

```python
y = np.array([0.8, 0.7, 0.9, 1.2, 1.4, 2.0, 1.5, 0.8])
F = np.array([0.6, 0.6, 0.6, 1.4, 1.5, 1.8, 1.4, 0.8])
S_1 = np.diag(1 / (0.1 * y)**2)
# S_1[3, 5] = 100
chi2 = (1/len(y)) * np.matmul(y-F, np.matmul(S_1, y-F))
print('chi^2 = {:.2f}'.format(chi2))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(S_1, annot=True, fmt=".2f", cmap="viridis", cbar=True)
plt.title("Annotated Heatmap of S_1")
plt.xlabel("Index")
plt.ylabel("Index")
plt.show()
```

```python
S = np.linalg.inv(S_1)
# Create the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(S, annot=True, fmt=".2f", cmap="viridis", cbar=True)
plt.title("Annotated Heatmap of S_1")
plt.xlabel("Index")
plt.ylabel("Index")
plt.show()
```
