```python
import xarray as xr
import matplotlib.pyplot as plt
```

```python
cvm_file = '/data/keeling/a/johntl3/b/LundstromPub2025/CAMP2Ex_RF17/intermediate_data/CVM_default_open.nc'
cvm = xr.load_dataset(cvm_file)
savefigs = True
```

```python
ticks = [0.0, 4.4, 8.8, 13.2, 17.6]
# facetgrid = cvm.rbg_mask.isel(x=slice(4,64,4)).plot.imshow(x='y', y='z', col='x', col_wrap=5)
facetgrid = cvm.rbg_mask.isel(x=[4, 16, 32, 48]).plot.imshow(x='y', y='z', col='x', col_wrap=2)

for ax in facetgrid.axes.flatten():
    ax.set_xticks(ticks)
facetgrid.fig.set_size_inches(10, 8)

if(savefigs):
    plt.savefig(
        '/data/keeling/a/johntl3/b/LundstromPub2025/CAMP2Ex_RF17/all_figures/V02_figure07_CVM.png',
        format='png',
        dpi=300,
        bbox_inches='tight',
        transparent=False
    )
```
