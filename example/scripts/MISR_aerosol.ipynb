```python
wavelength = 0.67
mixture = 21
aod = 0.2
leg_ord = 20

rte_grid_title = 'default'
rte_grid_file = '../data/RTG_{}.nc'.format(rte_grid_title)
rte_grid = xr.load_dataset(rte_grid_file)

aod_str = '{:.3f}'.format(aod).replace('.','p')

out_file = '../data/AEROSOL_MISR_MIX{}_AOD{}_LO{}_RTG_{}.nc'.format(mixture, aod_str, leg_ord, rte_grid_title)
```
