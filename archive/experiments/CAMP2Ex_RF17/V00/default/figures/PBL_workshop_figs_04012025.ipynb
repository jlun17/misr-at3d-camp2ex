```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
```

```python
ext3d_file = './extinction3D_notitle_movie_03/iter{:03d}.png'
obs_file = './obs_static/obs.png'
forward_file = './forward_sim/iter{:03d}.png'
chi2_file = './chi2_movie/iter{:03d}.png'

out_dir = './at3d_animation01/'
os.makedirs(out_dir, exist_ok=True)

# List of PNG files and their positions
for i in range(19):
    png_files = [
        {"file": ext3d_file.format(i), "x": 2.0, "y": 3.8, "width": 5.7},
        {"file": chi2_file.format(i+1), "x": 7.5, "y": 4.0, "width": 6.0},
        {"file": forward_file.format(i), "x": 2.0, "y": 0.1, "width": 12.0},
        {"file": obs_file, "x": 2.0, "y": 1.751, "width": 12.0},
    ]

    # Create a blank figure
    fig, ax = plt.subplots(figsize=(16, 9))
    ax.axis('off')  # Turn off axes

    # Add each PNG file to the figure
    for png in png_files:
        img = mpimg.imread(png["file"])
        height = png["width"] * img.shape[0] / img.shape[1]  # Maintain aspect ratio
        ax.imshow(img, extent=[png["x"], png["x"] + png["width"], png["y"], png["y"] + height])

    ax.set_xlim([0, 16])
    ax.set_ylim([0, 9])
    ax.set_title('Iteration: {}'.format(i), fontsize=32)
    # Save or show the figure
    plt.savefig(out_dir+'iter{:03d}'.format(i), dpi=300, bbox_inches='tight')
    plt.close(fig)
    # plt.show()
```
