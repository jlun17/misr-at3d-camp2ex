```python
def plot_rays_projected(ax, xlocs, mu, phi, plane='xz', kwargs=None):
    if(kwargs == None):
        kwargs = {'c':'tab:orange', 'lw': 0.8}
    if(plane=='xz'):
        theta_proj = np.arctan(np.tan(np.arccos(mu)) * np.cos(phi))
    elif(plane=='yz'):
        theta_proj = np.arctan(np.tan(np.arccos(mu)) * np.sin(phi))
    for x0 in xlocs:
        m = 1/np.tan(-theta_proj)
        x = np.array(list(ax.get_xlim()))
        y = m*x - m*x0
        ax.plot(x, y, **kwargs)
```
