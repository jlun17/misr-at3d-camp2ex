```python
# conda env spec
from pathlib import Path
import xarray as xr

PROJECT_ROOT = Path().resolve().parent.parent.parent
```

```python
RAW_DATA = PROJECT_ROOT / 'data' / 'raw'
era_mean = xr.load_dataset(RAW_DATA / 'era5' / 'ERA5_surface_CAMP2Ex_RF17_wide_ensemble_mean.nc')
era_spread = xr.load_dataset(RAW_DATA / 'era5' / 'ERA5_surface_CAMP2Ex_RF17_wide_ensemble_spread.nc')
era = xr.load_dataset(RAW_DATA / 'era5' / 'ERA5_surface_CAMP2Ex_RF17_wide.nc')
# data/raw/era5/ERA5_surface_CAMP2Ex_RF17_wide_ensemble_mean.nc
# data/raw/era5/ERA5_surface_CAMP2Ex_RF17_wide_ensemble_spread.nc
# data/raw/era5/ERA5_surface_CAMP2Ex_RF17_wide.nc
```
