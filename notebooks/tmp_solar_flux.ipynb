```python
import at3d
import xarray as xr
import numpy as np
```

# SZA = 60 degrees

```python
#load atmosphere
atmosphere = xr.open_dataset('./AFGL_summer_mid_lat.nc')

rte_grid = at3d.grid.make_grid(
    delx = 0.2,
    npx = 5,
    dely = 0.2,
    npy = 5,
    z = np.arange(0, 1.01, 0.1)
)
rayleigh_scattering = at3d.rayleigh.to_grid([0.67],atmosphere,rte_grid)
rayleigh_scattering[0.67]['extinction'] = (['x','y','z'], np.zeros((5, 5, 11)))

solver = at3d.solver.RTE(
    numerical_params = at3d.configuration.get_config(),
    source = at3d.source.solar(0.67, 0.5, 0, solarflux=500.0),
    surface = at3d.surface.lambertian(1),
    medium = {'null': rayleigh_scattering[0.67]},
    num_stokes = 1
)
solvers = at3d.containers.SolversDict()
solvers.add_solver(0.67, solver)

sensor_dict = at3d.containers.SensorsDict()
sensor = at3d.sensor.orthographic_projection(0.67, rte_grid, 0.2, 0.2, 0, 0, altitude='TOA', stokes='I')
sensor_dict.add_sensor('MISR', sensor)

sensor_dict.get_measurements(solvers)

print(np.unique(sensor_dict['MISR']['sensor_list'][0].I))
```
