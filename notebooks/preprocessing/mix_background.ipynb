```python
# conda env 'at3d_pure_k9'
"""This notebook merges two atmospheres using at3d.medium.mix_optical_properties

Generates MIXTURE*.nc
"""

from collections import OrderedDict
import at3d
import xarray as xr
import numpy as np
```

```python
## INPUT
aerosol_file = '/data/keeling/a/johntl3/b/LundstromPub2025/data/interim/AEROSOL_MISR_54_0p200.nc'
atmos_file = '/data/keeling/a/johntl3/b/LundstromPub2025/data/interim/ATMOS_CAMP2Ex_RF17_default.nc'

## OUTPUT
out_file = '/data/keeling/a/johntl3/b/LundstromPub2025/data/interim/MIXTURE_default_mix54.nc'
```

```python
# wavelength = 0.67

aerosol = xr.load_dataset(aerosol_file)
atmos = xr.load_dataset(atmos_file)

assert np.array_equal(aerosol.x.data, atmos.x.data)
assert np.array_equal(aerosol.y.data, atmos.y.data)
assert np.array_equal(aerosol.z.data, atmos.z.data)
```

```python
background_optical_scatterers = OrderedDict()
merged_properties = at3d.medium.mix_optical_properties(
    atmos,
    aerosol,
    asymmetry_tol=0.02,
    maxnphase=40
)

ones = np.ones((atmos.dims['x'], atmos.dims['y'], atmos.dims['z']))
if((np.array_equal(atmos.ssalb.data, ones)) & (np.array_equal(aerosol.ssalb.data, ones))):
    merged_properties['ssalb'] = (['x', 'y', 'z'], ones)

merged_properties.to_netcdf(out_file)
```
