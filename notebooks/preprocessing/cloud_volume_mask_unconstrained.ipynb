```python
"""This notebook generates an unconstrained cloud volume mask

Generates CVM*_unconstrained.nc
"""

import xarray as xr
import numpy as np

from pathlib import Path
PROJECT_ROOT = Path().resolve().parent.parent
INTERIM_DATA = PROJECT_ROOT / 'data' / 'interim'
```

```python
## INPUT
# RT Grid
rte_grid_file = INTERIM_DATA / 'RTG_default.nc'

## OUTPUT
out_file = INTERIM_DATA / 'CVM_CAMP2Ex_RF17_default_unconstrained.nc'
```

```python
rte_grid = xr.load_dataset(rte_grid_file)
```

```python
mask = np.ones((rte_grid.dims['x'], rte_grid.dims['y'], rte_grid.dims['z']))
mask[0,:,:] = mask[-1,:,:] = mask[:,0,:] = mask[:,-1,:] = 0.0
rte_grid['cloud_volume_mask'] = (['x','y','z'], mask)
print('Number of free parameters: {}'.format(np.sum(rte_grid.cloud_volume_mask.data)))
rte_grid.to_netcdf(out_file)
```
