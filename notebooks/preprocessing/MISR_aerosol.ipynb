```python
# conda env 'at3d_pure_k9'
"""This notebook generates aerosol optical properties given an 
aerosol optical depth, and MISR mixture model.

Generates AEROSOL_MISR*.nc
"""

import sys
sys.path.append('/data/keeling/a/johntl3/b/LundstromPub2025')

import src.preprocess.MISR_aerosol_mixture as asol
import xarray as xr
import numpy as np
import at3d
```

```python
## INPUT
wavelength = 0.67
aod = 0.200
mixture = 54
leg_ord = 166

rte_grid_file = '/data/keeling/a/johntl3/b/LundstromPub2025/data/interim/RTG_default.nc'

## OUTPUT
aod_str = '{:.3f}'.format(aod).replace('.', 'p')
out_file = '/data/keeling/a/johntl3/b/LundstromPub2025/data/interim/AEROSOL_MISR_{}_{}.nc'.format(mixture, aod_str)
print(out_file)
```

```python
rte_grid = xr.load_dataset(rte_grid_file)
```

```python
aerosol_optical_properties = asol.get_MISR_mixture_optical_properties(
    band=wavelength,
    mixture=mixture,
    aod=aod,
    rte_grid=rte_grid,
    acp_file='../../data/raw/MISR/MISR_AM1_ACP_MIXTURE_F06_0022.nc',
    legendre_order=leg_ord,
    g_tol=1e-2,
    res_tol=1
)
aerosol_optical_properties.to_netcdf(out_file)
print('Aerosol properties saved to: {}'.format(out_file))
```
