```python
# conda env 'at3d_pure_k9'
from pathlib import Path
PROJECT_ROOT = Path().resolve().parent

import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
```

```python
lut_file = PROJECT_ROOT / 'NK1990_1D_Retrieval' / 'LUT_I_V05_periodic.nc'
LUT = xr.load_dataset(lut_file)
```

```python
f, ax = plt.subplots(1, 1, figsize=(7,5))
LUT.tau.plot(y='I', ax=ax)
s = ax.scatter(LUT.tau, LUT.I, c=LUT.num_unique)
ax.set_xscale('log')
plt.colorbar(s, label='Num Unique')
```
